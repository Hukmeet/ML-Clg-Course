```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 1. Load the SMS Spam Collection dataset from the public repository
url = "https://raw.githubusercontent.com/justmarkham/pycon-2016-tutorial/master/data/sms.tsv"

# -------------------------------------------------------------------
# STEP 1: Load and inspect the dataset
# -------------------------------------------------------------------
# Read the TSV file containing message labels and text content
df = pd.read_csv(url, sep='\t', header=None, names=['label', 'message'])

print(f"Successfully loaded {len(df)} rows of data from dataset!\n")
print(df.head())

# -------------------------------------------------------------------
# STEP 2: Preprocess text and vectorize using TF-IDF
# -------------------------------------------------------------------
# Convert string labels into binary numeric values (ham: 0, spam: 1)
df['label_num'] = df.label.map({'ham': 0, 'spam': 1})

# Split the dataset into training and testing subsets
X_train, X_test, y_train, y_test = train_test_split(df['message'], df['label_num'], test_size=0.2, random_state=42)

# Initialize TF-IDF Vectorizer to extract numerical features from text
vectorizer = TfidfVectorizer(stop_words='english')
X_train_vectorized = vectorizer.fit_transform(X_train)
X_test_vectorized = vectorizer.transform(X_test)

# -------------------------------------------------------------------
# STEP 3: Train model and evaluate performance
# -------------------------------------------------------------------
# Train a Multinomial Naive Bayes classifier
model = MultinomialNB()
model.fit(X_train_vectorized, y_train)

# Predict and calculate accuracy score on test data
predictions = model.predict(X_test_vectorized)
accuracy = accuracy_score(y_test, predictions)

print(f"\nModel Training Complete! Test Accuracy: {accuracy * 100:.2f}%")

# Test the trained model with sample messages
sample_messages = [
    "Free entry in 2 a wkly comp to win FA Cup final tkts",
    "Hey, are we still meeting for lunch today?"
]
sample_vectorized = vectorizer.transform(sample_messages)
sample_preds = model.predict(sample_vectorized)

print("\nSample Predictions:")
for msg, pred in zip(sample_messages, sample_preds):
    label = "Spam" if pred == 1 else "Ham"
    print(f"Message: '{msg}' --> Prediction: {label}")
```
