```python
import pandas as pd
import requests

# 1. Type any city name here! (e.g., "Dewas", "Indore", "Bhopal")
city_name = "Dewas"

# -------------------------------------------------------------------
# STEP 1: Get coordinates using Open-Meteo's Geocoding API
# -------------------------------------------------------------------
geocoding_url = f"https://geocoding-api.open-meteo.com/v1/search?name={city_name}&count=1&language=en&format=json"

geo_response = requests.get(geocoding_url).json()

# Extract latitude and longitude from the top result
first_result = geo_response["results"][0]
latitude = first_result["latitude"]
longitude = first_result["longitude"]

print(
    f"Found coordinates for '{city_name}': Latitude {latitude}, Longitude {longitude}"
)

# -------------------------------------------------------------------
# STEP 2: Fetch weather data using your preferred parameters
# -------------------------------------------------------------------
# We insert the dynamic latitude & longitude into your custom link:
weather_url = (
    f"https://api.open-meteo.com/v1/forecast?"
    f"latitude={latitude}&longitude={longitude}&"
    f"hourly=temperature_2m,relative_humidity_2m,precipitation_probability,rain,cloud_cover&"
    f"timezone=auto"
)

weather_response = requests.get(weather_url).json()

# -------------------------------------------------------------------
# STEP 3: Convert the response into a Pandas DataFrame and save as CSV
# -------------------------------------------------------------------
# Extract the 'hourly' dictionary (contains time, temperature, rain, etc.)
hourly_data = weather_response["hourly"]

# Build the DataFrame
df = pd.DataFrame(hourly_data)

# Save to CSV using the city name in the filename
csv_filename = f"{city_name.lower()}_weather.csv"
df.to_csv(csv_filename, index=False)

print(f"Successfully saved {len(df)} rows of data to '{csv_filename}'!\n")
print(df.head())  # Shows the first 5 rows
```
